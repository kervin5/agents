```python
# Below is a list of workflow patterns:
# 1. Prompt chaining: A series of prompts where the output of one is used as input to another
# 2. Routing: An LLM makes the decision about which model to use based on the question
# 3. Paralleization: Multiple models are queried simultaneously to be aggregated
# 4. Orchestrator worker: Similar to parallelization, but a model is used for orchestration and synthesis of the responses
# 5. Evaluator-optimizer: A model is used to evaluate: aceptance, rejection, or ranking of responses
```

```python
# We're going to to build a workflow that tells you what to wear depending on the weather and your gender. It'll be a fashion expert.
# We'll use the orchestrator worker pattern, where we have a model that decides which model to use based on the question.

openai = OpenAI()




def determine_outfit_for_male(weather):
    prompt = f"""You are a fashion expert. Based on the weather conditions described below, please provide a detailed outfit recommendation
    for a male.
    Weather: {weather}
    """

    messages = [{"role": "user", "content": prompt}]

    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
    )
    answer = response.choices[0].message.content

    return answer


def determine_outfit_for_female(weather):
    prompt = f"""You are a fashion expert. Based on the weather conditions described below, please provide a detailed outfit recommendation
    for a female. The attire should be stylish and appropriate for the weather as well as stand out in a crowd.
    Weather: {weather}
    """
    messages = [{"role": "user", "content": prompt}]

    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
    )
    answer = response.choices[0].message.content

    return answer

def determine_outfit(weather, gender):
    if gender == "male":
        return determine_outfit_for_male(weather)
    else:
        return determine_outfit_for_female(weather)

outfit = determine_outfit("sunny and 25 degrees Celsius", "male")

print("Outfit recommendation:", outfit)
```

```python
# Modern OpenAI API with structured tools

def male_fashion_expert(weather):
    """Male fashion specialist"""
    prompt = f"""You are a male fashion expert. Based on the weather conditions: {weather}
    Provide a detailed, stylish outfit recommendation for men."""
    
    messages = [{"role": "user", "content": prompt}]
    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
    )
    return response.choices[0].message.content

def female_fashion_expert(weather):
    """Female fashion specialist"""
    prompt = f"""You are a female fashion expert. Based on the weather conditions: {weather}
    Provide a detailed, stylish outfit recommendation for women that stands out."""
    
    messages = [{"role": "user", "content": prompt}]
    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
    )
    return response.choices[0].message.content

def general_fashion_expert(weather):
    """General fashion specialist"""
    prompt = f"""You are a general fashion expert. Based on the weather conditions: {weather}
    Provide a versatile outfit recommendation suitable for anyone."""
    
    messages = [{"role": "user", "content": prompt}]
    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
    )
    return response.choices[0].message.content

FASHION_TOOLS = {
    "male_fashion_expert": male_fashion_expert,
    "female_fashion_expert": female_fashion_expert,
    "general_fashion_expert": general_fashion_expert
}

def orchestrator_agent(weather, gender):
    """
    Modern orchestrator agent using OpenAI's structured tool calling
    """
    tools = [
        {
            "type": "function",
            "function": {
                "name": "select_fashion_specialist",
                "description": "Selects the appropriate fashion specialist based on weather and gender",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "specialist": {
                            "type": "string",
                            "enum": ["male_fashion_expert", "female_fashion_expert", "general_fashion_expert"],
                            "description": "The specialist to use for fashion recommendations"
                        },
                        "reasoning": {
                            "type": "string",
                            "description": "Brief explanation for the selection"
                        }
                    },
                    "required": ["specialist"]
                }
            }
        }
    ]
    
    messages = [
        {
            "role": "user", 
            "content": f"Select the appropriate fashion specialist for weather: {weather}, gender: {gender}"
        }
    ]
    
    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
        tools=tools,
        tool_choice={"type": "function", "function": {"name": "select_fashion_specialist"}}
    )
    
    tool_call = response.choices[0].message.tool_calls[0]
    function_args = json.loads(tool_call.function.arguments)
    return function_args["specialist"]

def autonomous_outfit_recommendation(weather, gender):
    """
    Updated main function using modern tool calling
    """
    # Get tool recommendation from orchestrator
    selected_tool = orchestrator_agent(weather, gender)
    
    # Execute the selected tool
    if selected_tool in FASHION_TOOLS:
        return FASHION_TOOLS[selected_tool](weather)
    else:
        return general_fashion_expert(weather)

# Test the updated system
outfit_recommendation = autonomous_outfit_recommendation("sunny and 25 degrees Celsius", "male")
print("Modern autonomous outfit recommendation:", outfit_recommendation)
```

```python
# Traditional autonomous tool selection approach

def male_fashion_expert(weather):
    """Male fashion specialist"""
    prompt = f"""You are a male fashion expert. Based on the weather conditions: {weather}
    Provide a detailed, stylish outfit recommendation for men."""
    
    messages = [{"role": "user", "content": prompt}]
    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
    )
    return response.choices[0].message.content

def female_fashion_expert(weather):
    """Female fashion specialist"""
    prompt = f"""You are a female fashion expert. Based on the weather conditions: {weather}
    Provide a detailed, stylish outfit recommendation for women that stands out."""
    
    messages = [{"role": "user", "content": prompt}]
    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
    )
    return response.choices[0].message.content

def general_fashion_expert(weather):
    """General fashion specialist"""
    prompt = f"""You are a general fashion expert. Based on the weather conditions: {weather}
    Provide a versatile outfit recommendation suitable for anyone."""
    
    messages = [{"role": "user", "content": prompt}]
    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
    )
    return response.choices[0].message.content

FASHION_TOOLS = {
    "male_fashion_expert": male_fashion_expert,
    "female_fashion_expert": female_fashion_expert,
    "general_fashion_expert": general_fashion_expert
}

def orchestrator_agent(weather, gender):
    """
    Traditional orchestrator agent using text-based decision making
    """
    prompt = f"""You are a fashion coordination assistant. Given the weather conditions "{weather}" and gender preference "{gender}", 
    you need to select the most appropriate fashion specialist.

    Available specialists:
    - male_fashion_expert: Best for male-specific fashion advice
    - female_fashion_expert: Best for female-specific fashion advice  
    - general_fashion_expert: Best for universal fashion advice

    Respond with ONLY the specialist name that would be most appropriate. No explanation needed."""
    
    messages = [{"role": "user", "content": prompt}]
    
    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
    )
    
    selected_tool = response.choices[0].message.content.strip()
    
    # Validate the response and fallback if needed
    if selected_tool not in FASHION_TOOLS:
        # Simple rule-based fallback
        if gender.lower() == "male":
            return "male_fashion_expert"
        elif gender.lower() == "female":
            return "female_fashion_expert"
        else:
            return "general_fashion_expert"
    
    return selected_tool

def autonomous_outfit_recommendation(weather, gender):
    """
    Traditional autonomous approach using text-based tool selection
    """
    # Get tool recommendation from orchestrator
    selected_tool = orchestrator_agent(weather, gender)
    
    # Execute the selected tool
    return FASHION_TOOLS[selected_tool](weather)

# Test the traditional system
outfit_recommendation = autonomous_outfit_recommendation("sunny and 25 degrees Celsius", "male")
print("Traditional autonomous outfit recommendation:", outfit_recommendation)
```
